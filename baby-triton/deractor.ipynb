Python Decorators
In Python, a decorator is a design pattern that allows you to modify the functionality of a function by wrapping it in another function.

The outer function is called the decorator, which takes the original function as an argument and returns a modified version of it.

```python

def jit(fn):
    def inner():
        print("jit is called")
    return inner

@jit
def add():
    print("add")

add()
```

```python
def jit(fn):
    def inner():
        print("jit is called")
    return inner

def add():
    print("add")

jit(add)()
```

```python
def jit(fn):
    def inner():
        print("jit is called")
    return fn

def add():
    print("add")

jit(add)()
```

下面这例子请注意

当你使用 @jit(target="cpu") 时，Python 会先调用 jit(target="cpu")。

这一步返回的是 inner 函数。

相当于执行了：decorator = jit(target="cpu")。

接下来，Python 会将 add 作为参数传递给 inner 函数。

相当于执行了：add = decorator(add)，即 add = inner(add)。

inner 函数接收 add 作为参数 fn，但是并未返回函数

当你调用 add() 会报错

```python
def jit(target="cpu"):
    def inner(fn):
        print("jit is called")
    return inner

@jit(target="cpu")
def add():
    print("add")

#add
```

上面的add这一行注释即时去掉也是一样的结果 add是None

```python
def jit(target="cpu"):
    def inner(fn):
        print("jit is called")
    return inner

def add():
    print("add")

decorator = jit(target="cpu")
add = decorator(add)
add() #此时产生了报错
```
